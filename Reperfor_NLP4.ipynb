```python
print('Arif Raini - UIN 231A026')
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
import string

nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('stopwords')

lemmatizer = WordNetLemmatizer()
```

```python
text = "The runner is running fast."
```

```python
print('Arif Raini - UIN 231A026')
text_lower = text.lower()
print(text_lower)
```

```python
print('Arif Raini - UIN 231A026')
tokens = word_tokenize(text_lower)
print(tokens)
```

```python
print('Arif Raini - UIN 231A026')
text_clean1 = [token for token in tokens if token not in string.punctuation]
print(text_clean1)
```

```python
print('Arif Raini - UIN 231A026')
lemmatizer = WordNetLemmatizer()
text_clean2 = [lemmatizer.lemmatize(token, pos = 'v') for token in text_clean1]
print(text_clean2)
```

```python
print('Arif Raini - UIN 231A026')
lemmatizer.lemmatize('catcher', pos = 'n')
```

```python
print('Arif Raini - UIN 231A026')

for token, lemma in zip(tokens, text_clean2):
    print(f'Token: {token} | Lemma: {lemma}')
```

```python
print('Arif Raini - UIN 231A026')
morphological_info = {}
for token, lemma in zip(tokens, text_clean2):
    if token != lemma:
      if token.endswith("ing"):
        analysis = f"Inflectional: Present participle/continuous form of '{lemma}'."
      elif token.endswith("er"):
        analysis = f"Derivational: '{lemma}' + er -> person who does an action."
      else:
        analysis = f"Inflectional or irregular form of '{lemma}'."
    else:
      analysis = "Base/Root form."
    morphological_info[token] = analysis

morphological_info
```

```python
print('Arif Raini - UIN 231A026')
print("Original text:", text)
print("Lowercase text:", text_lower)
print("Tokenized text:", tokens)
print("Cleaned text:", text_clean1)
print("Lemmatized text:", text_clean2)
```
