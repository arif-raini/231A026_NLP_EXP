**Name:** Arif Raini  
**UIN:** 231A026

```python
print("Name: Arif Raini | UIN: 231A026")
# Import Libraries & Download Resources
import string
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
print("Name: Arif Raini | UIN: 231A026")
# Original Text
text = "The Misadventures of Penny (TMOP) is a semi-serialized, character-driven slice-of-life crime dramedy. The target audience is young adults and upward. In this world where all humans and non-human toons coexist, is Dimen City, a bustling metropolis where people from all walks of life can be met. One such is Penny McCartney, a sharp-tongued, quick-witted legal arms dealer and gunslinger navigating her life along with her best friend Walter Sterling, a tech genius who keeps things logical, and their shared robot companion, 8. Given the crime ridden nature of Dimen City, it is not uncommon to find criminal activity and gangs, the largest of which is The Hooligans. Having committed almost every known crime under the sun, The Hooligans are the vile, scum-of-the-earth, irredeemable monsters. When Penny and her team happen to break one of The Hooligans' major trafficking operations, they find themselves under the crime syndicate's radar and have to balance between work, life, and involuntary vigilantism."
print(text)
```

```python
print("Name: Arif Raini | UIN: 231A026")
# Text to Lowercase
text_lower = text.lower()
print(text_lower)
```

```python
print("Name: Arif Raini | UIN: 231A026")
punctuations = list(string.punctuation)
print(punctuations)
```

```python
print("Name: Arif Raini | UIN: 231A026")
# Tokenized Text
tokenized_text = word_tokenize(text_lower)
print(tokenized_text)
```

```python
print("Name: Arif Raini | UIN: 231A026")
# Removing Punctuation
cleaned_text1 = []
for word in tokenized_text:
  if word not in punctuations:
    cleaned_text1.append(word)
print(cleaned_text1)
```

```python
print("Name: Arif Raini | UIN: 231A026")
print(stopwords.words('english'))
print(len(stopwords.words('english')))
```

```python
print("Name: Arif Raini | UIN: 231A026")
# Removing Stopwords
cleaned_text2 = []
for word in cleaned_text1:
  if word not in stopwords.words('english'):
    cleaned_text2.append(word)
print(cleaned_text2)
```

```python
print("Name: Arif Raini | UIN: 231A026")
# Stemming
stemmer = PorterStemmer()
stemmed_text = []
for word in cleaned_text2:
  stemmed_text.append(stemmer.stem(word))
print(stemmed_text)
```

```python
print("Name: Arif Raini | UIN: 231A026")
# Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_text = []
for word in cleaned_text2:
  lemmatized_text.append(lemmatizer.lemmatize(word))
print(lemmatized_text)
```

```python
print("Name: Arif Raini | UIN: 231A026")
print("Original Text:")
print(text)
print("\nStemmed Text:")
print(stemmed_text)
print("Doug was here...")
print("\nLemmatized Text:")
print(lemmatized_text)
```
